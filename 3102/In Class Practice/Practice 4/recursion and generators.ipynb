```python
def find_files(fs_dict, ext, path=''):
    found_files = []
    for key, value in fs_dict.items():
        current_path = path + '/' + key if path else key
        if isinstance(value, dict):
            found_files.extend(find_files(value, ext, current_path))
        elif key.endswith(ext):
            found_files.append(current_path)
    return found_files
```

```python
def get_errors(log_path):
    with open(log_path, 'r') as log_file:
        for line in log_file:
            if line.startswith('ERROR'):
                yield line
```

3.1 Because the file system has a tree structure. Each folder can contain files and subfolders, and recursion allows the same logic to be applied repeatedly to each subfolder until all files have been explored.

3.2 Recursion requires one function call for each log entry. With millions of log lines, this would create a very deep call stack, consume large amounts of memory, and may exceed Python’s recursion limit.
