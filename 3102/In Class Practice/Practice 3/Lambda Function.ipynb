```python
# Sigmoid probabilities
import math
logits = [-2.2, -1.4, 0.8, 2.0]
probabilities = list(map(lambda x: 1/ (1 + math.exp(-x)), logits))
print(probabilities)
```

```python
# Filter predictions
predictions = [{"label": "cat", "conf": 0.1},
               {"label": "dog", "conf": 0.99},
               {"label": "rabbit", "conf": 0.95},
               {"label": "hamster", "conf": 0.05}]
filtered_predictions = list(filter(lambda x:x["conf"] >= 0.9, predictions))
print(filtered_predictions)
```

```python
# Rank object detection results
detections = [{"label": "cat", "conf": 0.1, "box": [10, 20, 30, 40]},
              {"label": "dog", "conf": 0.99, "box": [50, 60, 70, 80]},
              {"label": "dog", "conf": 0.95, "box": [90, 100, 110, 120]},
              {"label": "cat", "conf": 0.05, "box": [130, 140, 150, 160]}]
sorted_detections = sorted(detections, key=lambda d: (d["conf"], (d["box"][2] - d["box"][0]) * (d["box"][3] - d["box"][1])), reverse=True)
print(sorted_detections)
```

```python
# Rewrite 'Sigmoid probabilities' using list comprehension
import math
logits = [-2.2, -1.4, 0.8, 2.0]
probabilities = [1 / (1 + math.exp(-x)) for x in logits]
print(probabilities)
```

The list comprehension version is clearer because it is easier to read and directly shows the calculation applied to each item.
