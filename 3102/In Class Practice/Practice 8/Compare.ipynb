```python
from transformers import AutoTokenizer, AutoModelForCausalLM
model_name = "Qwen/Qwen2-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
prompts = [
    "Artificial intelligence will change the future because",
    "A long time ago in a galaxy far, far away,",
    "The most important lesson I learned in college was"
]
```

```python
for prompt in prompts:
    inputs = tokenizer(prompt, return_tensors="pt")
    print(f"Prompt: {prompt}")

    top_k_output = model.generate(**inputs, do_sample=True, top_k=50, top_p=1.0, num_return_sequences=3, max_new_tokens=50)
    print("Top-k Sampling (top_k=50):")
    for i, output in enumerate(top_k_output, start=1):
        generated_text = tokenizer.decode(output, skip_special_tokens=True)
        print(f"Output {i}: {generated_text}")
    
    top_p_output = model.generate(**inputs, do_sample=True, top_k=0, top_p=0.9, num_return_sequences=3, max_new_tokens=50)
    print("Top-p Sampling (top_p=0.9):")
    for i, output in enumerate(top_p_output, start=1):
        generated_text = tokenizer.decode(output, skip_special_tokens=True)
        print(f"Output {i}: {generated_text}")

    print()
```

## Comparison Analysis

### Diversity

Both top-k and top-p sampling produced different continuations for each
prompt, showing that both methods introduced diversity into the generated
text. For the artificial intelligence prompt, the outputs discussed
different aspects such as automation, productivity, machine learning,
risks, and industrial applications. For the story prompt, the outputs
developed substantially different settings, characters, and civilizations.
The top-p outputs sometimes followed more unexpected directions, while the
top-k outputs were generally more focused on the original topic.

### Coherence

For the artificial intelligence prompt, both methods produced mostly
coherent explanations, although some outputs ended before completing their
ideas because of the token limit. The college prompt also produced generally
logical continuations under both methods. The story prompt showed a larger
difference: several outputs were coherent narratives, but some contained
illogical combinations or sudden changes in topic. For example, one top-p
output stated that a village contained two villages, which reduced its
coherence. Overall, top-k appeared slightly more coherent in these samples.

### Repetition

Neither sampling method showed severe repetition across most outputs.
However, some individual outputs repeated words or sentence structures.
For example, the top-p story output repeatedly used the word “village” in
the sentences “there was a village” and “The village had two villages.”
The artificial intelligence outputs also repeated common terms such as
“AI,” “technology,” and “fields,” but these repetitions were mostly natural
because they were relevant to the topic. There was no clear overall winner
in terms of repetition.

### Relevance

Most outputs remained relevant to their prompts. The artificial intelligence
outputs continued discussing the effects and applications of AI, while the
college outputs generally described personal lessons or learning experiences.
The story outputs remained broadly related to distant galaxies and fictional
civilizations, although some introduced unrelated fictional names and mixed
elements from different stories. In this experiment, top-k outputs were
slightly more consistently relevant, while top-p occasionally moved in more
unexpected directions.
