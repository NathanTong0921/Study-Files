```python
import sys
record_list = ["2026001", "Alice", 95]
record_tuple = ("2026001", "Alice", 95)
record_dict = {"student_id": "2026001", "name": "Alice", "score": 95}
```

### 1. Function total_size(obj)

```python
def total_size(obj):
    total = sys.getsizeof(obj)
    if isinstance(obj, dict):
        for key, value in obj.items():
            total += sys.getsizeof(key) + sys.getsizeof(value)
    else:
        for item in obj:
            total += sys.getsizeof(item)
    return total
```

```python
list_size = total_size(record_list)
tuple_size = total_size(record_tuple)
dict_size = total_size(record_dict)

print("Memory usage:")
print("List size:", list_size, "bytes")
print("Tuple size:", tuple_size, "bytes")
print("Dict size:", dict_size, "bytes")
```

### 2. Comparison of the Three Representations

#### 1. List
- Memory usage: Medium
- Readability: Not very good because we need to remember the meaning of each index.
- Mutability: Mutable. We can change its elements.

#### 2. Tuple
- Memory usage: Low
- Readability: Not very good because we need to remember the meaning of each index.
- Mutability: Immutable. We cannot change its elements after it is created.

#### 3. Dictionary
- Memory usage: High
- Readability: Good because every value has a meaningful key.
- Mutability: Mutable. We can change its values.

### 3. Choosing a Representation for 10,000 Student Records

I would choose **tuple**. This is because tuple uses less memory than a list or dictionary, so it is suitable for storing a large number of fixed records.

#### However, the best choice also depends on the requirements:

If the records need to be changed, I would choose list. This is because list is mutable and uses less memory than dictionary.

If readability and convenience are more important than memory usage, I would choose dictionary.

### 4. Better representation: 

```python
# Use a class with __slots__
class Student:
    __slots__ = ["student_id", "name", "score"]

    def __init__(self, student_id, name, score):
        self.student_id = student_id
        self.name = name
        self.score = score

    def __str__(self):
        return f"Student ID: {self.student_id}, Name: {self.name}, Score: {self.score}"
```

```python
student = Student("2026001", "Alice", 95)
student_total_size = (
    sys.getsizeof(student)
    + sys.getsizeof(student.student_id)
    + sys.getsizeof(student.name)
    + sys.getsizeof(student.score)
)
print("Total student size:", student_total_size, "bytes")
```

```python
# Modify the score
student.score = 100
print(student)
```

### Why this representation is better:
- It is more readable than list and tuple, because we can use student.name and student.score.
- It is mutable, because we can change student.score.
- It usually uses less memory than dictionary, because `__slots__` avoids storing a per-object dictionary.
